# PCA with Housing Dataset

## Context
Consider the situation where you are working for Zillow as a data scientist.

Housing pricing predictions is the goal.

We know 80 things about each house to use as inputs to be able to predict the price of a
house.

Your goal is to islotate the important features from the dataset and build a model which
can be used to predict the price of the houses. ‘

Since there are too many features, Principal Component Analysis can be applied to
reduce the number of features used for the actual prediction model, without any loss of
information,

https://www.kaggle.com/c/home-data-for-ml-course/data?select=train.csv

- Download the train.csv file and load it using pandas

With 79 explanatory variables describing (almost) every aspect of residential homes in
Ames, Iowa, this competition challenges you to predict the final price of each home.

```python
#As a backup assigning to new variable
df1 = df 
df1.shape
df1.info()
```

```python
#Standardizing the continuous data
from sklearn.preprocessing import StandardScaler 
scale = StandardScaler()
df1_cat = df1.select_dtypes(include=object) #Filtering object/ categorical variables
df1_cat.info() #15 variables
df1_cat.columns #Categorical variables names
```

```python
df1_con = df1.select_dtypes(exclude=object) #Filtering non object/ categorical variables
df1_con.info() #14 variables in including target variable
df1_con = df1_con.drop('SalePrice', axis=1) #dropping target variable
df1_con.info() #13 Predictors
df1_con.columns
```

```python
df1_tar = pd.DataFrame(df1.SalePrice) #Target variable
df1_tar.info()
```

```python
#Scaling the continuous variables without target variable
df1_con = pd.DataFrame(scale.fit_transform(df1_con), columns = df1_con.columns)
```

```python
pca = PCA(n_components=13)
df_pca = pca.fit_transform(df1_con)
```

```python
#Array to dataframe
df_pca = pd.DataFrame(df_pca) #Principal components
```

```python
pvctr_13 = pca.components_ #Eigen Vectors

pevl_13 = pca.explained_variance_ #Eigen values
#The amount of variance explained by each of the selected components.
```

```python
pca.explained_variance_ratio_
#Percentage of variance explained by each of the selected components.
```

```python
#Cummulative sum of explained_variance_ratio_
np.cumsum(pca.explained_variance_ratio_)
```

```python
# Plot of Cummulative sum of explained_variance_ratio_
plt.plot(np.cumsum(pca.explained_variance_ratio_), color= 'r')
plt.title('Cummulative - PCA Explained Variance Ratio')
plt.xlabel('PCA n_components')
plt.ylabel('PCA- explained_variance_ratio')
plt.xticks(np.arange(0,14,1))
plt.yticks(np.arange(0.3,1.1,0.05))
plt.grid(color='b', linestyle='dotted', linewidth=.5)
plt.show()
```

```python
#Selecting 5 PCA's
df_pca_5 = df_pca.iloc[:, :5]
df_pca_5.columns = ['PCA0', 'PCA1', 'PCA2', 'PCA3', 'PCA4']
df_pca_5.info()
```

```python
#Concatenating 15 categorical variables, 5 PCA's and Target variable
#Resetting the index
df1_cat = df1_cat.reset_index(drop=True)
df_pca_5 = df_pca_5.reset_index(drop=True)
df1_tar = df1_tar.reset_index(drop=True)
```

```python
#Concatenating 3 dataframes
df_pca1 = pd.concat([df1_cat,df_pca_5,df1_tar], axis=1)
df_pca1.info()
```

```python
df_pca1.columns
```

```python
#Building a linear regression model
from statsmodels.formula.api import ols
model1 = ols('''SalePrice~MSZoning+LotShape+Neighborhood+Condition1+BldgType+
             RoofStyle+MasVnrType+ExterQual+BsmtQual+BsmtExposure+BsmtFinType1
             +HeatingQC+KitchenQual+GarageType+GarageFinish+PCA0+PCA1+PCA2+PCA3
             +PCA4''', data=df_pca1).fit()
             
model1.summary()
```
