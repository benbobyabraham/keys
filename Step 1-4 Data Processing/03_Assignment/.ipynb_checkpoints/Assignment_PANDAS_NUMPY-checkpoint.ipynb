## 2) Run the given code and find the dot product of both arrays.

```python
a = np.random.randint(2,10, (3,4))
b = np.random.randint(2,10, (4,3))
print(a)
print(b)
```

```python
a.dot(b)
```

```python
b.dot(a)
```

```python
data.OBJECTID=data.OBJECTID.mask(data.OBJECTID==0).fillna(data.OBJECTID.mean()).astype(int)
print(data)
```
